### 9. Compute the moving average of a series with a window of 3.

```python
# your answer here
```

**One possible solution:**

```python
s = np.array([1., 2., 3., 4., 5., 6., 7.])

# option 1: cumsum trick
c = np.cumsum(np.insert(s, 0, 0))
print((c[3:] - c[:-3]) / 3)

# option 2: convolution
print(np.convolve(s, np.ones(3) / 3, mode='valid'))
```

### 11. Compute the Euclidean distance between every pair of rows in a (5, 3) array. The result should be (5, 5) with zeros on the diagonal.

```python
# your answer here
```

**One possible solution:**

```python
rng = np.random.default_rng(4)
P = rng.random((5, 3))
D = np.sqrt(((P[:, None, :] - P[None, :, :]) ** 2).sum(axis=-1))
print(D.round(3))
print(np.allclose(D, D.T), np.allclose(np.diag(D), 0))
```

### 14. Solve this system:  2x + y - z = 8,  -3x - y + 2z = -11,  -2x + y + 2z = -3

```python
# your answer here
```

**One possible solution:**

```python
A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)
x = np.linalg.solve(A, b)
print(x)    # expect [2, 3, -1]
```

### 15. Estimate π by sampling random points in a unit square and counting how many fall inside the quarter circle.

```python
# your answer here
```

**One possible solution:**

```python
rng = np.random.default_rng(5)
pts = rng.random((1_000_000, 2))
inside = (pts ** 2).sum(axis=1) <= 1
print(4 * inside.mean())
```

### 16. Simulate 10,000 games of rolling two dice and estimate the probability of each total (2 to 12).

```python
# your answer here
```

**One possible solution:**

```python
rng = np.random.default_rng(6)
totals = rng.integers(1, 7, size=(10_000, 2)).sum(axis=1)
probs = np.bincount(totals, minlength=13)[2:] / len(totals)
for t, p in zip(range(2, 13), probs):
    print(f'{t:2d}: {p:.3f}  ' + '#' * int(p * 200))
```

### 17. Implement one step of the Game of Life on a small grid, using only array operations.

```python
# your answer here
```

**One possible solution:**

```python
def life_step(g):
    # count the 8 neighbours by summing shifted copies of the grid
    n = sum(np.roll(np.roll(g, dy, 0), dx, 1)
            for dy in (-1, 0, 1) for dx in (-1, 0, 1)
            if (dy, dx) != (0, 0))
    return ((n == 3) | ((g == 1) & (n == 2))).astype(int)

glider = np.zeros((6, 6), dtype=int)
glider[0, 1] = glider[1, 2] = glider[2, 0] = glider[2, 1] = glider[2, 2] = 1

g = glider
for step in range(3):
    print(g, end='\n\n')
    g = life_step(g)
```
